```python
import yfinance as yf 
```

```python
start_date = "2024-06-25"
end_date = "2025-06-25"

# Stock symbol
stock_symbol = "TATAMOTORS.NS"
```

```python
stock_data = yf.download(
    tickers=stock_symbol,
    start=start_date,
    end=end_date,
    interval="1d",
    actions=True,
    progress=False,
    multi_level_index=False
)
```

```python
stock_data[stock_data["Dividends"] > 0]
```
